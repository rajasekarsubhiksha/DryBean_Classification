```python
# Separate independent and dependent variables

X = df.drop("Class", axis=1)
y = df["Class"]

print("Input Features Shape:", X.shape)
print("Target Variable Shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Features:", X_train.shape)
print("Testing Features:", X_test.shape)

print("Training Target:", y_train.shape)
print("Testing Target:", y_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
```

```python
# Fit scaler using training data only
X_train_scaled = scaler.fit_transform(X_train)

# Apply the same scaling to testing data
X_test_scaled = scaler.transform(X_test)

print("Training data after scaling:", X_train_scaled.shape)
print("Testing data after scaling:", X_test_scaled.shape)
```

```python
print("Mean of scaled training features:")
print(np.round(X_train_scaled.mean(axis=0), 2))

print("\nStandard deviation of scaled training features:")
print(np.round(X_train_scaled.std(axis=0), 2))
```

Model 1 - Logistic Regression 

```python
from sklearn.linear_model import LogisticRegression

print("Logistic Regression imported successfully!")
```

```python
# Create Logistic Regression model

lr_model = LogisticRegression(
    max_iter=2000,
    random_state=42
)

# Train the model using scaled training data

lr_model.fit(X_train_scaled, y_train)

print("Logistic Regression model trained successfully!")
```

```python
# Predict using testing data

y_pred_lr = lr_model.predict(X_test_scaled)

print("Predictions generated successfully!")

print("\nFirst 10 Actual Values:")
print(y_test[:10].values)

print("\nFirst 10 Predicted Values:")
print(y_pred_lr[:10])
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    matthews_corrcoef
)

# Calculate evaluation metrics

accuracy = accuracy_score(y_test, y_pred_lr)

precision = precision_score(
    y_test, y_pred_lr, average='weighted'
)

recall = recall_score(
    y_test, y_pred_lr, average='weighted'
)

f1 = f1_score(
    y_test, y_pred_lr, average='weighted'
)

# Calculate multi-class ROC-AUC
y_prob_lr = lr_model.predict_proba(X_test_scaled)

roc_auc = roc_auc_score(
    y_test,
    y_prob_lr,
    multi_class='ovr',
    average='weighted',
    labels=lr_model.classes_
)

mcc = matthews_corrcoef(y_test, y_pred_lr)

# Display results

print("LOGISTIC REGRESSION - EVALUATION RESULTS")
print("-----------------------------------------")
print("Accuracy :", round(accuracy, 4))
print("Precision:", round(precision, 4))
print("Recall   :", round(recall, 4))
print("F1 Score :", round(f1, 4))
print("ROC-AUC  :", round(roc_auc, 4))
print("MCC      :", round(mcc, 4))
```

```python
from sklearn.metrics import confusion_matrix, classification_report
import matplotlib.pyplot as plt
import seaborn as sns

# Generate confusion matrix
cm_lr = confusion_matrix(y_test, y_pred_lr, labels=lr_model.classes_)

# Display confusion matrix
plt.figure(figsize=(9, 6))

sns.heatmap(
    cm_lr,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=lr_model.classes_,
    yticklabels=lr_model.classes_
)

plt.title("Logistic Regression - Confusion Matrix")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.show()
```

```python
print("LOGISTIC REGRESSION - CLASSIFICATION REPORT")
print(classification_report(y_test, y_pred_lr))
```

Model 5 – Support Vector Machine (SVM)

```python

from sklearn.svm import SVC
from sklearn.calibration import CalibratedClassifierCV

# Create SVM model with probability calibration
svm_model = CalibratedClassifierCV(
    estimator=SVC(
        kernel="rbf",
        random_state=42
    ),
    method="sigmoid",
    cv=5,
    ensemble=False
)

# Train the model
svm_model.fit(X_train_scaled, y_train)

print("Support Vector Machine trained successfully!")
```

```python

# Generate predictions
y_pred_svm = svm_model.predict(X_test_scaled)

print("SVM predictions generated successfully!")

print("\nActual Values:")
print(y_test.iloc[:10].values)

print("\nPredicted Values:")
print(y_pred_svm[:10])
```

```python

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    matthews_corrcoef
)

# Calculate evaluation metrics
accuracy_svm = accuracy_score(y_test, y_pred_svm)

precision_svm = precision_score(
    y_test, y_pred_svm, average="weighted"
)

recall_svm = recall_score(
    y_test, y_pred_svm, average="weighted"
)

f1_svm = f1_score(
    y_test, y_pred_svm, average="weighted"
)

# Multiclass ROC-AUC
y_prob_svm = svm_model.predict_proba(X_test_scaled)

auc_svm = roc_auc_score(
    y_test,
    y_prob_svm,
    multi_class="ovr",
    average="weighted",
    labels=svm_model.classes_
)

mcc_svm = matthews_corrcoef(y_test, y_pred_svm)

# Display results
print("SUPPORT VECTOR MACHINE - PERFORMANCE METRICS")
print("---------------------------------------------")
print("Accuracy:", round(accuracy_svm, 4))
print("Precision:", round(precision_svm, 4))
print("Recall:", round(recall_svm, 4))
print("F1 Score:", round(f1_svm, 4))
print("ROC-AUC:", round(auc_svm, 4))
print("MCC:", round(mcc_svm, 4))
```

```python

from sklearn.metrics import confusion_matrix, classification_report
import matplotlib.pyplot as plt
import seaborn as sns

# Generate confusion matrix
cm_svm = confusion_matrix(
    y_test,
    y_pred_svm,
    labels=svm_model.classes_
)

# Plot confusion matrix
plt.figure(figsize=(9, 6))

sns.heatmap(
    cm_svm,
    annot=True,
    fmt="d",
    cmap="Reds",
    xticklabels=svm_model.classes_,
    yticklabels=svm_model.classes_
)

plt.title("SVM - Confusion Matrix")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.show()

# Generate classification report
print("SUPPORT VECTOR MACHINE - CLASSIFICATION REPORT")

print(
    classification_report(
        y_test,
        y_pred_svm
    )
)
```
